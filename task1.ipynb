```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("messy_data.csv")
num_cols = df.select_dtypes(include="number").columns
num_cols
```

```python
df[num_cols].describe()
```

```python
summary = pd.DataFrame({
    "Mean": df[num_cols].mean(),
    "Median": df[num_cols].median(),
    "Std": df[num_cols].std(),
    "Min": df[num_cols].min(),
    "Max": df[num_cols].max(),
    "25%": df[num_cols].quantile(0.25),
    "50%": df[num_cols].quantile(0.50),
    "75%": df[num_cols].quantile(0.75)
})

summary
```

```python
range_values = df[num_cols].max() - df[num_cols].min()
range_values.sort_values(ascending=False)
```

```python
# identify identical columns
cat_cols = df.select_dtypes(include="object").columns
cat_cols
df[cat_cols].describe()
```

The numerical columns were identified and their statistical summary was generated using describe().

Mean, median, standard deviation, minimum, maximum, and percentile values were analyzed.

The range between minimum and maximum values was compared to identify columns with large variations.
