```python
import pandas as pd

df = pd.read_csv("messy_data.csv")
```

```python
df["Expected_Total"] = df["Unit_Price"] * df["Quantity"] * (1 - df["Discount"] / 100)
df["Difference"] = df["Total_Amount"] - df["Expected_Total"]

df[["Unit_Price", "Quantity", "Discount", "Total_Amount", "Expected_Total", "Difference"]].head()
```

```python
df["Difference"].abs().sort_values(ascending=False).head(10)
```

```python
df.loc[df["Difference"].abs().nlargest(10).index]
```

The expected order amount was calculated using unit price, quantity, and discount. 

The calculated values were compared with the existing total amount to identify records with large differences that may require further investigation.
