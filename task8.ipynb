```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv("messy_data.csv")
```

```python
columns = ["Age", "Unit_Price", "Quantity", "Discount", "Rating", "Delivery_Days", "Total_Amount"]

for col in columns:
    plt.figure(figsize=(8, 4))
    sns.boxplot(x=df[col])
    plt.title(col)
    plt.show()
```

```python
for col in columns:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    count = ((df[col] < lower) | (df[col] > upper)).sum()
    print(col, ":", count)
```

The boxplots were used to visually identify possible outliers in the numerical columns.

 Extreme values were observed in some columns and should be investigated before deciding whether they are errors or legitimate values.
