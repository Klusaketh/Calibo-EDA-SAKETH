```python
import pandas as pd

df = pd.read_csv("messy_data.csv")
```

```python
df.isnull().sum()
```

```python
missing = df.isnull().sum()
missing[missing > 0]
```

```python
missing = df.isnull().sum()
missing[missing > 0]
```

```python
for col in df.select_dtypes(include="number").columns:
    df[col] = df[col].fillna(df[col].median())

for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
df.isnull().sum()
```

Numerical missing values are replaced with the median, while categorical missing values are replaced with the mode. This avoids removing useful records from the dataset.

Duplicate records were identified using duplicated(), and duplicate Order_ID values were checked separately. Duplicate records can affect analysis by counting the same transaction more than once and producing incorrect results.
