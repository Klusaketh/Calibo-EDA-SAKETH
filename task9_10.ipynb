```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv("messy_data.csv")
```

```python
columns = ["Age", "Unit_Price", "Total_Amount"]

for col in columns:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    outliers = df[(df[col] < lower) | (df[col] > upper)]

    print(col)
    print("Q1:", q1)
    print("Q3:", q3)
    print("IQR:", iqr)
    print("Lower Bound:", lower)
    print("Upper Bound:", upper)
    print("Number of Outliers:", len(outliers))
    print()
```

```python
for col in columns:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr

    print("\n", col)
    print(df[(df[col] < lower) | (df[col] > upper)].head())
```

The IQR method was used to identify potential outliers in three numerical columns. 

Q1, Q3, IQR, lower bound, and upper bound were calculated, and values outside these limits were treated as potential outliers.

```python
import pandas as pd
df = pd.read_csv("messy_data.csv")
```

```python
columns = ["Age", "Unit_Price", "Total_Amount"]

for col in columns:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr

    outliers = df[(df[col] < lower) | (df[col] > upper)]

    print(col)
    print("Outliers:", len(outliers))
    print(outliers[[col]].head(10))
    print()
```

```python
print("Age: Extreme values should be checked for unrealistic ages.")
print("Unit_Price: High values may be legitimate product prices.")
print("Total_Amount: High values may be legitimate large orders.")
df = df[(df["Age"] >= 10) & (df["Age"] <= 100)]
```

```python
print("Rows after cleaning:", len(df))
print(df["Age"].min(), df["Age"].max())
```

The identified outliers were reviewed before removal instead of treating every extreme value as an error.

 Only clearly unrealistic age values were removed, while extreme prices and order amounts were retained because they may represent            legitimate transactions.
