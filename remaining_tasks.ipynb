Task 17 — Correlation Analysis

```python
corr = df.select_dtypes(include="number").corr()
corr
```

```python
total_corr = corr["Total_Amount"].sort_values(ascending=False)
total_corr
```

```python
strongest_positive = corr.where(corr < 1).stack().sort_values(ascending=False).head(1)
strongest_negative = corr.stack().sort_values().head(1)

print("Strongest Positive Correlation:")
print(strongest_positive)

print("\nStrongest Negative Correlation:")
print(strongest_negative)
```

The correlation matrix was used to understand the relationships between numerical variables.

 Variables with higher correlation with Total_Amount may be more useful for understanding or predicting the total order value.

```python
print("Most ordered category:", df["Product_Category"].value_counts().idxmax())
print("Most popular payment method:", df["Payment_Method"].value_counts().idxmax())
print("City with highest sales:", df.groupby("City")["Total_Amount"].sum().idxmax())
print("Highest average order category:", df.groupby("Product_Category")["Total_Amount"].mean().idxmax())
print("Highest average spending gender:", df.groupby("Gender")["Total_Amount"].mean().idxmax())
print("Return percentage:", df["Returned"].value_counts(normalize=True).get("Yes", 0) * 100)
print("Highest rated product:", df.groupby("Product")["Rating"].mean().idxmax())
print("Highest average delivery days:", df.groupby("Product_Category")["Delivery_Days"].mean().idxmax())
print("Highest average quantity category:", df.groupby("Product_Category")["Quantity"].mean().idxmax())
print("Highest average discount category:", df.groupby("Product_Category")["Discount"].mean().idxmax())
```
