```python
import pandas as pd

df = pd.read_csv("messy_ecommerce_15000_student_practice.csv")
```

```python
cat_cols = df.select_dtypes(include="object").columns
cat_cols
```

```python
df[cat_cols].describe()
```

```python
df[cat_cols].nunique()
```

```python
df[cat_cols].apply(lambda x: x.value_counts().iloc[0])
```

```python
print("Most common Gender:", df["Gender"].mode()[0])
print("Most common City:", df["City"].mode()[0])
print("Most common Product Category:", df["Product_Category"].mode()[0])
print("Most common Payment Method:", df["Payment_Method"].mode()[0])
print("Most common Return Status:", df["Returned"].mode()[0])
```

The categorical columns were identified and summarized using Pandas.

Unique values, most frequent values, and their frequencies were calculated.

The most common values for Gender, City, Product Category, Payment Method, and Return Status were identified.
